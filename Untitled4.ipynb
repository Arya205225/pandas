```python
# group Aggregation 
import numpy as np 
import pandas as pd 
```

```python
# group by 
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2026-01-01', periods=8)
}
df=pd.DataFrame(data)
```

```python
df
```

```python
# group by categary and calculate the sum of sales 
s= df.groupby('Category')['Sales'].sum()
s
```

```python
# group by store and calculate the sum of sales 
s = df.groupby('Store')["Sales"].sum()
s
```

```python
# group by multiple columns 
# group by category 
x = df . groupby(['Category','Store'])['Sales'].sum()
```
